## 1. Scenario
A synthetic bank loan portfolio for Malaysian retail customers. The goal is to predict whether a customer will **default** on a loan, using financial attributes and a free-text customer feedback comment.

## 2. Target variable
| Column | Description |
|---|---|
| `default` | 1 = customer defaulted, 0 = did not default. Target rate is about 22%, so the classes are moderately imbalanced. |

## 3. Schema
| Column | Type | Description |
|---|---|---|
| `customer_id` | ID | Unique ID (C0001–C1200) |
| `age` | Numeric | 21–65 |
| `annual_income` | Numeric | RM 24,000–300,000 |
| `loan_amount` | Numeric | RM 5,000–500,000 |
| `loan_term_months` | Numeric | 12–360 |
| `credit_score` | Numeric | 300–850 |
| `existing_debts` | Numeric | RM 0–200,000 |
| `employment_years` | Numeric | 0–40 |
| `employment_type` | Categorical | Salaried / Self-employed / Contract / Unemployed |
| `region` | Categorical | Klang Valley / Penang / Johor / Sabah / Sarawak / Other |
| `loan_purpose` | Categorical | Home / Car / Education / Business / Personal |
| `customer_feedback` | Text | 1–3 sentences in the customer's own voice (used by Member 2 for LLM feature extraction) |
| `default` | Target | 0 / 1 |

## 4. Generation method
- **Tool / model:** [e.g. Claude Sonnet 5.5]
- **Process:** The dataset was generated in 12 batches of 100 records (C0001–C1200) using a fixed prompt template. Only the starting customer ID changed between batches. Batches were saved in `data/raw/batches/` and merged below.
- **How the data was shaped:** Default probability was made more likely with a low credit score, high debt relative to income, a short employment history, and contract or unemployed status, with random noise so it is not perfectly predictable. Feedback tone loosely matches financial stress, with some deliberate mismatches.
- **Prompt log:** The full prompts are in `reports/prompts_member1.md`.
